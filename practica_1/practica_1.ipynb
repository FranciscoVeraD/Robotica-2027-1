## Introducción

En el desarrollo de proyectos de robótica es necesario utilizar diferentes herramientas de software para **crear archivos, automatizar tareas, documentar procedimientos y compartir el trabajo**. Aunque estas herramientas tienen funciones diferentes, su combinación permite establecer flujos de trabajo reproducibles y facilita la colaboración.

- <img src="imagenes/Ubuntu-logo.png" height=60/> **Ubuntu** es un sistema operativo basado en Linux ampliamente utilizado en robótica. El manejo de la terminal permite ejecutar programas, administrar archivos y automatizar tareas sin depender exclusivamente de interfaces gráficas.

- **Bash (Bourne Again SHell)** es un intérprete de comandos utilizado habitualmente en sistemas Linux. Además de permitir ejecutar comandos de manera interactiva, permite crear **scripts de Bash**, es decir, archivos de texto que contienen una secuencia de comandos que puede ejecutarse de manera automatizada.

- <img src="imagenes/Jupyter-logo.png" height=60/> **Jupyter** proporciona un entorno interactivo basado en notebooks. Un notebook está formado por celdas que pueden contener texto, ecuaciones, código y resultados. En esta práctica se utilizará para documentar el trabajo y ejecutar comandos de Bash desde el mismo documento.

- **Markdown** es un lenguaje de marcado ligero que permite dar formato al texto mediante una sintaxis sencilla. Jupyter utiliza Markdown para crear títulos, listas, tablas, enlaces y ecuaciones dentro de las celdas de texto.

- <img src="imagenes/Git-logo.png" height=60/> **Git** es un sistema de control de versiones que permite registrar cambios en los archivos de un proyecto, recuperar versiones anteriores y trabajar mediante ramas.

- <img src="imagenes/Github-logo.png" height=60/> **GitHub** es una plataforma que permite alojar repositorios Git en línea y facilita el trabajo colaborativo mediante repositorios, ramas, colaboradores y herramientas de revisión.

En robótica, estas herramientas pueden utilizarse conjuntamente. Por ejemplo, un alumno puede crear un script de Bash para preparar una carpeta de trabajo, documentar el procedimiento en Jupyter y finalmente almacenar el notebook y los scripts en un repositorio de GitHub.


## Cuestionario previo

### Responder de forma breve las siguientes preguntas:

- En Linux, ¿qué es una **shell** y qué relación existe entre `sh` y `bash`?
> Un shell es un programa que actúa como puente entre el usuario y el núcleo de un sistema operativo. Sh es el modelo original de Bash, por su parte, Bash es la versión moderna que incluye todo lo que tenía Sh, además de nuevas funciones.

- ¿Qué es un **script de Bash** y cuál es la diferencia entre ejecutar un comando directamente en la terminal y ejecutarlo mediante un script?
> Un script de Bash es un archivo de texto plano que contiene una lista de comandos ordenados para que el shell los ejecute uno tras otro de forma automática.
Ejecutar un comando en la terminal es una acción manual e inmediata, y un script es una acción automatizada y reutilizable.

- ¿Qué es **Jupyter** y qué tipo de información puede contener una celda de un notebook?
> Jupyter es un entorno interactivo que permite crear y visualizar archivos de Jupyter Notebook de forma estética. Por medio de GitHub, el notebook se convierte en un archivo fácil de leer.
Puede contener celdas de código (Code), celdas de texto con formato (Markdown) y celdas de texto sin formato (Raw).

- ¿Qué es **Markdown** y para qué se utiliza dentro de Jupyter?
> Es un lenguaje de marcado ligero que sirve para aplicar formato a un texto utilizando símbolos simples y fáciles de leer.
Se utiliza para convertir un archivo simple de texto en un reporte o documento interactivo, con títulos, listas, imágenes y ecuaciones.

- ¿Cuál es la diferencia entre **Git** y **GitHub**?
> Git es la herramienta de software que rastrea los cambios en tus archivos.
GitHub es la plataforma en la nube que almacena tus archivos y permite compartirlos con otros.

- La forma general de la matriz de transformación homogénea de un sistema $j$ respecto a un sistema $i$.
$$
{}^iT_j = \begin{bmatrix}
R_{3 \times 3} & P_{3 \times 1} \\
0_{1 \times 3} & 1
\end{bmatrix} = \begin{bmatrix}
r_{11} & r_{12} & r_{13} & p_x \\
r_{21} & r_{22} & r_{23} & p_y \\
r_{31} & r_{32} & r_{33} & p_z \\
0 & 0 & 0 & 1
\end{bmatrix}
$$

- La forma general de la matriz de rotación de un sistema $j$ respecto a un sistema $i$ respecto al eje $z$.
$$
{}^iR_j(z, \theta) = \begin{bmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

- La forma general del vector de posición de un sistema $j$ respecto a un sistema $i$.
$$
{}^iP_j = \begin{bmatrix}
p_x \\

\end{bmatrix}

$$

## Análisis de resultados

Responder las siguientes preguntas con base en el trabajo realizado:

1. ¿Qué ventajas ofrece un script de Bash frente a ejecutar manualmente una secuencia de comandos en la terminal?

> En la práctica, crear la estructura de Practica1 a mano implicaba teclear mkdir, varios touch, tree y rm -r uno por uno. Con el script practica1_bash.sh todo se hace con una sola ejecución, siempre en el mismo orden y sin errores de tecleo. El script también permitió agregar lógica que a mano no se tiene: con un if verifica si Practica1 ya existe y, si es así, se detiene sin borrar nada, y usa $HOME para funcionar en la computadora de cualquier integrante. Además, queda guardado y versionado en el repositorio, así que cualquiera puede repetir exactamente el mismo procedimiento.


2. ¿Qué diferencias existen entre una celda de Markdown y una celda de código en Jupyter?

> En este notebook usamos las celdas de Markdown para documentar: los títulos de cada sección, las instrucciones, las imágenes (como la captura de la ejecución del script, insertada desde imagenes/), el código del script y la explicación de cada bloque. Esas celdas solo se renderizan, no se ejecutan. Una celda de código, en cambio, contiene instrucciones que sí se ejecutan y muestran su salida debajo, por ejemplo al correr comandos de Bash desde el notebook. Con ambas, el notebook funciona como un reporte que reúne la explicación, el código y los resultados en un solo archivo.

3. ¿Qué utilidad tiene un repositorio en línea para almacenar scripts, notebooks y documentación de un proyecto de robótica?

> En esta práctica, el repositorio Robotica-2027-1 reúne en un solo lugar el script (scripts/), el notebook (practica_1.ipynb), las imágenes y el README.md con los integrantes y la organización de las prácticas. Al agregar a los integrantes como colaboradores, todos pueden subir y descargar cambios sin pasarse archivos por otros medios, y Git guarda el historial de commits para ver qué cambió y volver atrás si algo se rompe. El .gitignore evita subir archivos innecesarios como .ipynb_checkpoints/. En un proyecto de robótica, donde el código, los notebooks y la documentación crecen con cada práctica, esto da orden, respaldo y la posibilidad de clonar y reproducir el trabajo en otra computadora.
